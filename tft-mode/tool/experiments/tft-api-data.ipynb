```python
import os
import json
import pandas as pd

current_folder = "\\".join(os.getcwd().split("\\")[:-1])
data_folder = f"{current_folder}\\data"


def loadFromJson(filePath):
    if filePath is None:
        print(f"File path {filePath} not exist!")
        return None

    with open(filePath, "r") as f:
        data = json.load(f)
    return data
```

```python
data = loadFromJson(os.path.join(data_folder, "champion.json"))

# Extract the relevant part of the data
champion_data = data['data']
```

```python
# Create a DataFrame from the extracted data
df = pd.DataFrame(champion_data)
```
