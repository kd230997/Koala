Basic OpenCV

```python
import os
import cv2 as cv
from matplotlib import pyplot as plt
import numpy as np

current_folder = "\\".join(os.getcwd().split("\\")[:-1])
folder_images = os.path.join(current_folder, "images")
img_path = os.path.join(folder_images, "template", "template.png")
my_img = cv.imread(img_path, cv.IMREAD_COLOR)
if my_img is None:
    print("Image not found")
else:
    print("Image found")
    plt.imshow(my_img)
    plt.show()
```
